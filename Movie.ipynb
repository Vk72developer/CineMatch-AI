```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):

  # Lowercase
  text = str(text).lower()

  # Remove the Punctuation
  text = re.sub(r'[^\w\s]', '', text)

  words = text.split()

  # remove stopwords
  words = [word for word in words if word not in stop_words]

  # lemmatize
  words = [lemmatizer.lemmatize(word) for word in words]

  return " ".join(words)
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index = df['title']).drop_duplicates()
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000, stop_words='english', ngram_range=(1, 2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python

def recommend(title, n=10):

    if title not in indices:
        return ['Movie not found']

    # Get movie index
    idx = indices[title]

    # Calculate cosine similarity
    sim_score = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    # Get top N similar movie indices
    similaridx = sim_score.argsort()[::-1][1:n+1]

    # Return recommended movie titles
    return df['title'].iloc[similaridx].tolist()
```

```python
recommend('Toy Story')
```

```python
recommend('Jumanji', 5)
```
